```python
import re
import pandas as pd
import glob

your_name = "Dhruti"

all_messages = []

files = glob.glob("*.txt")

print("Files found:", files)

for file_name in files:

    with open(file_name, "r", encoding="utf-8") as file:
        lines = file.readlines()

    for line in lines:

        # Remove invisible unicode characters
        line = line.replace("\u200e", "").strip()

        # WhatsApp pattern
        pattern = r"^\[(.*?)\] (.*?): (.*)"

        match = re.match(pattern, line)

        if match:

            sender = match.group(2)
            message = match.group(3)

            # Keep only your messages
            if your_name.lower() in sender.lower():

                # Skip media messages
                if "omitted" in message.lower():
                    continue

                # Remove links
                message = re.sub(r"http\S+", "", message)

                # Remove extra spaces
                message = message.strip()

                # Ignore tiny useless replies
                if len(message) > 2:
                    all_messages.append(message)

print("Total cleaned messages:", len(all_messages))
```

```python
df = pd.DataFrame(all_messages, columns=["message"])

df.to_csv("cleaned_chat_dataset.csv", index=False)

print("Dataset saved successfully!")
```
